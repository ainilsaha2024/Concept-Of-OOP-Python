```python
class  student_id:
    def __init__(self,name,Id):
        self.name = name 
        self.Id = Id 
        print("Wellcome to oop in python")
    
```

```python
class my_fast:
    def __init__(self,city,age,number):
        self.city = city
        self.age = age
        self.number = number 
        print ("How are you ?")

c =my_fast("Dhaka",22,1011)
a =my_fast("narsingdi",23,1012)
n =my_fast("Raipura",24,1013) 

n.number = 9999

print()

print(f"This is c in city {c.city}") 
print(f"This is a in age {a.age}")
print(f"This is n in number {n.number}") 
```

```python
class school:
    def __init__(self,city,age,number):
        self.city = city
        self.age = age
        self.number = number 
 
    def details(self): # instance method 
        print("City:",self.city ," " ,"age:",self.age ," ","number:",self.number)

x =school("Dhaka",22,1011)
y =school("narsingdi",23,1012)
z =school("Raipura",24,1013)
y.city="Narsingdi"
        
```

```python
x.details()
y.details()
z.details()
```
