# Real life OOP Projact ATM 

```python
class AtmMachine:
    def __init__(self):
        self.pin = ""
        self.balance = 0 
        self.menu()

    def menu(self):
        user_input = input(
            """
Hi, how can I help you today?
1. Press 1 to create a new PIN Number.
2. Press 2 to change your PIN Number.
3. Press 3 to check your balance.
4. Press 4 to withdraw money.
5. Press anything else to exit and logout.
=> """
        )

        if user_input == "1":
            self.create_pin()
        elif user_input == "2":
            self.change_pin()
        elif user_input == "3":
            self.check_balance()
        elif user_input == "4":
            self.withdraw_money()
        else:
            print("Thank you for using our ATM Machine. GOOD BYE!")
            exit()

    def create_pin(self):
        user_pin = input("Enter your 4-digit new PIN number: ")
        if len(user_pin) == 4 and user_pin.isdigit():
            self.pin = user_pin
            try:
                user_balance = float(input("Enter your initial balance: "))
                self.balance = user_balance
                print("PIN created and balance set successfully.")
            except ValueError:
                print("Invalid balance amount.")
        else:
            print("PIN must be exactly 4 digits and numeric.")
        self.menu()

    def change_pin(self):
        old_pin = input("Enter your current PIN: ")
        if old_pin == self.pin:
            new_pin = input("Enter your new 4-digit PIN: ")
            if len(new_pin) == 4 and new_pin.isdigit():
                self.pin = new_pin
                print("PIN changed successfully.")
            else:
                print("New PIN must be 4 digits and numeric.")
        else:
            print("Incorrect current PIN.")
        self.menu()

    def check_balance(self):
        entered_pin = input("Enter your PIN to check balance: ")
        if entered_pin == self.pin:
            print(f"Your current balance is: ${self.balance:.2f}")
        else:
            print("Incorrect PIN.")
        self.menu()

    def withdraw_money(self):
        entered_pin = input("Enter your PIN to withdraw money: ")
        if entered_pin == self.pin:
            try:
                amount = float(input("Enter amount to withdraw: "))
                if amount <= 0:
                    print("Amount must be greater than zero.")
                elif amount > self.balance:
                    print("Insufficient balance.")
                else:
                    self.balance -= amount
                    print(f"${amount:.2f} withdrawn successfully.")
                    print(f"New balance: ${self.balance:.2f}")
            except ValueError:
                print("Invalid amount.")
        else:
            print("Incorrect PIN.")
        self.menu()
```
